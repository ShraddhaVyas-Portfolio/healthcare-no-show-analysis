Key Findings
1. Appointment Attendance Rate
Approximately 79.81% of patients attended their appointments, while 20.19% did not attend. This indicates that nearly 1 in 5 appointments resulted in a no-show.

2. Age Influences Attendance
Patients who attended appointments had an average age of 37.79 years, compared to 34.32 years for patients who missed appointments. This suggests that younger patients were slightly more likely to miss appointments.

3. Scholarship Recipients Had Higher No-Show Rates
Patients receiving scholarships had a no-show rate of 23.74%, compared to 19.81% for patients without scholarships. This may indicate that socioeconomic factors influence appointment attendance.

4. SMS Reminders Did Not Reduce No-Shows
Patients who received SMS reminders had a no-show rate of 27.57%, while those who did not receive SMS reminders had a no-show rate of 16.70%.

This result is surprising and suggests that receiving an SMS reminder alone does not guarantee appointment attendance. Further investigation is needed to understand this relationship.

5. Chronic Conditions Were Slightly More Common Among Attendees
Patients who attended appointments had slightly higher rates of hypertension (20.44% vs 16.90%) and diabetes (7.38% vs 6.41%) than patients who missed appointments. This suggests that patients with ongoing health conditions may be more motivated to attend scheduled appointments.
